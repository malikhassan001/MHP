```python
import ipywidgets as widgets
from IPython.display import display, Audio, clear_output, FileLink
from google.colab import files

# Interface
text_box = widgets.Textarea(
    value="Hello! Welcome to my voice cloning tool.",
    placeholder="Yahan apna text paste karo...",
    description="Text:",
    layout=widgets.Layout(width="100%", height="150px")
)

language_box = widgets.Dropdown(
    options=[
        ("Hindi", "hi"),
        ("English", "en")
    ],
    value="hi",
    description="Language:"
)

generate_btn = widgets.Button(
    description="Generate Voice",
    button_style="success",
    icon="play"
)

output = widgets.Output()

def generate_voice(b):
    with output:
        clear_output()

        text = text_box.value.strip()
        language = language_box.value

        if not text:
            print("Please enter some text!")
            return

        print("Generating voice... Please wait.")

        try:
            output_file = "my_cloned_voice.wav"

            tts.tts_to_file(
                text=text,
                speaker_wav=reference_audio,
                language=language,
                file_path=output_file
            )

            print("Voice generated successfully!")

            display(Audio(output_file))
            display(FileLink(output_file))

        except Exception as e:
            print("Error:", str(e))

generate_btn.on_click(generate_voice)

display(text_box)
display(language_box)
display(generate_btn)
display(output)
```

```python
import os

BASE_DIR = "/content/drive/MyDrive/VoiceCloningTool"

INPUT_DIR = os.path.join(BASE_DIR, "Input")
CHUNKS_DIR = os.path.join(BASE_DIR, "Audio_Chunks")
OUTPUT_DIR = os.path.join(BASE_DIR, "Final_Audio")
TEMP_DIR = os.path.join(BASE_DIR, "Temp")

for folder in [INPUT_DIR, CHUNKS_DIR, OUTPUT_DIR, TEMP_DIR]:
    os.makedirs(folder, exist_ok=True)

print("All folders created successfully!")

print("Project location:", BASE_DIR)
```

```python
from google.colab import files
import os
import re
import json

print("Apni YouTube story ki TXT file upload karo:")
uploaded_script = files.upload()

script_filename = next(iter(uploaded_script))
script_path = os.path.join(INPUT_DIR, script_filename)

with open(script_path, "wb") as f:
    f.write(uploaded_script[script_filename])

with open(script_path, "r", encoding="utf-8-sig") as f:
    full_script = f.read()

# Clean empty spaces
full_script = re.sub(r'\n{3,}', '\n\n', full_script).strip()

# Split into manageable chunks
MAX_CHARS = 2000

sentences = re.split(r'(?<=[.!?।])\s+', full_script)

chunks = []
current_chunk = ""

for sentence in sentences:
    sentence = sentence.strip()

    if not sentence:
        continue

    while len(sentence) > MAX_CHARS:
        if current_chunk:
            chunks.append(current_chunk)
            current_chunk = ""

        chunks.append(sentence[:MAX_CHARS])
        sentence = sentence[MAX_CHARS:]

    if len(current_chunk) + len(sentence) + 1 <= MAX_CHARS:
        current_chunk += (" " if current_chunk else "") + sentence
    else:
        if current_chunk:
            chunks.append(current_chunk)
        current_chunk = sentence

if current_chunk:
    chunks.append(current_chunk)

# Save chunks
chunks_file = os.path.join(TEMP_DIR, "story_chunks.json")

with open(chunks_file, "w", encoding="utf-8") as f:
    json.dump(chunks, f, ensure_ascii=False, indent=2)

print("\nScript uploaded successfully!")
print("Total words:", len(full_script.split()))
print("Total characters:", len(full_script))
print("Total chunks:", len(chunks))
print("Chunk size:", MAX_CHARS, "characters")
```

```python
import os
import json
import shutil

# Save reference voice permanently in Google Drive
REFERENCE_PATH = os.path.join(BASE_DIR, "reference_voice.mp3")

if not os.path.exists(REFERENCE_PATH):
    shutil.copy2(reference_audio, REFERENCE_PATH)

# Save all chunks permanently
CHUNKS_JSON = os.path.join(BASE_DIR, "story_chunks.json")

with open(CHUNKS_JSON, "w", encoding="utf-8") as f:
    json.dump(chunks, f, ensure_ascii=False, indent=2)

# Create progress tracker
PROGRESS_FILE = os.path.join(BASE_DIR, "progress.json")

if not os.path.exists(PROGRESS_FILE):
    progress = {
        "total_chunks": len(chunks),
        "completed_chunks": [],
        "status": "not_started"
    }

    with open(PROGRESS_FILE, "w", encoding="utf-8") as f:
        json.dump(progress, f, indent=2)

print("Resume system setup successful!")
print("Reference voice saved!")
print("Story chunks saved!")
print("Progress tracker ready!")
print("Total chunks:", len(chunks))
```

```python
import os
import json
import torch
from tqdm.notebook import tqdm

PROGRESS_FILE = os.path.join(BASE_DIR, "progress.json")

# Load previous progress
with open(PROGRESS_FILE, "r", encoding="utf-8") as f:
    progress = json.load(f)

completed = set(progress["completed_chunks"])

progress["status"] = "running"

def save_progress():
    with open(PROGRESS_FILE, "w", encoding="utf-8") as f:
        json.dump(progress, f, indent=2)

save_progress()

print("Starting voice generation...")
print("Total chunks:", len(chunks))
print("Already completed:", len(completed))

for i, chunk in enumerate(tqdm(chunks, desc="Generating voice")):

    if i in completed:
        continue

    output_file = os.path.join(
        CHUNKS_DIR,
        f"chunk_{i+1:04d}.wav"
    )

    try:
        tts.tts_to_file(
            text=chunk,
            speaker_wav=REFERENCE_PATH,
            language="hi",
            file_path=output_file
        )

        completed.add(i)
        progress["completed_chunks"] = sorted(list(completed))
        save_progress()

    except Exception as e:
        print(f"\nError in chunk {i+1}: {e}")
        print("Progress saved. You can resume later.")
        break

if len(completed) == len(chunks):
    progress["status"] = "completed"
    save_progress()
    print("\nAll voice chunks generated successfully!")

else:
    print("\nCompleted:", len(completed), "/", len(chunks))
```

```python
!apt-get -qq install ffmpeg

import os
import subprocess

chunk_files = [
    os.path.join(CHUNKS_DIR, f"chunk_{i+1:04d}.wav")
    for i in range(len(chunks))
]

missing = [f for f in chunk_files if not os.path.exists(f)]

print("Total chunks:", len(chunk_files))
print("Missing chunks:", len(missing))

if not missing:
    with open(os.path.join(TEMP_DIR, "audio_list.txt"), "w") as f:
        for path in chunk_files:
            f.write(f"file '{path}'\n")

    final_audio = os.path.join(OUTPUT_DIR, "Hindi_Story_Final.wav")

    subprocess.run([
        "ffmpeg", "-y",
        "-f", "concat",
        "-safe", "0",
        "-i", os.path.join(TEMP_DIR, "audio_list.txt"),
        "-c:a", "pcm_s16le",
        final_audio
    ], check=True)

    print("Final audio created:", final_audio)
else:
    print("Some chunks are missing!")
```
